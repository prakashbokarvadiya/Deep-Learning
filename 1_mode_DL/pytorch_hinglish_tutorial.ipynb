# PyTorch Full Tutorial 
Is notebook me hum PyTorch ke basic se advanced methods seekhenge.

```python
# 1. Import torch
# PyTorch library import karte hain
import torch
```

```python
# 2. Tensor banana
# Tensor ek data structure hai (array jaisa)
x = torch.tensor([1, 2, 3])
print(x)
```

```python
# Zeros, Ones, Random tensor
print(torch.zeros(3))  # sab 0
print(torch.ones(3))   # sab 1
print(torch.rand(3))   # random values
```

```python
# 3. Shape check karna
print(x.shape)
```

```python
# 4. Mathematical operations
a = torch.tensor([1,2])
b = torch.tensor([3,4])

print(a + b)   # addition
print(a * b)   # multiplication
```

```python
# Matrix multiplication
m1 = torch.rand(2,2)
m2 = torch.rand(2,2)
print(torch.matmul(m1, m2))
```

```python
# 5. Autograd (gradient calculation)
x = torch.tensor(2.0, requires_grad=True)
y = x * x
y.backward()
print(x.grad)  # dy/dx
```

```python
# 6. Neural Network banana
import torch.nn as nn

model = nn.Linear(2,1)
print(model)
```

```python
# 7. Loss Function
loss_fn = nn.MSELoss()
```

```python
# 8. Optimizer
import torch.optim as optim
optimizer = optim.SGD(model.parameters(), lr=0.01)
```

```python
# 9. Training Loop
x = torch.randn(10,2)
y = torch.randn(10,1)

for epoch in range(5):
    optimizer.zero_grad()  # gradients reset
    output = model(x)      # forward pass
    loss = loss_fn(output, y)  # loss calculate
    loss.backward()        # backward pass
    optimizer.step()       # update weights
    print(f"Epoch {epoch}, Loss: {loss.item()}")
```
