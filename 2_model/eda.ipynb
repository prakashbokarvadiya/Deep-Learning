```python
import pandas as pd 
```

```python
data=pd.read_csv('student_ann_dataset.csv')
data.shape 
```

```python
data.head()
```

```python
x=data.drop(columns='result')
y=data['result']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
import torch
import torch.nn as nn
```

```python
# x_train (Pandas DataFrame) ko NumPy array me convert karke
# usse PyTorch tensor bana rahe hain (float32 type me)
x_train = torch.tensor(x_train.values, dtype=torch.float32)

# y_train (target values) ko bhi tensor me convert kar rahe hain
# .view(-1,1) ka matlab: usse column shape me convert karna (2D tensor)
# jaise: [1,0,1] → [[1],[0],[1]]
y_train = torch.tensor(y_train.values, dtype=torch.float32).view(-1,1)

# test input data ko bhi same tarike se tensor me convert kiya
x_test = torch.tensor(x_test.values, dtype=torch.float32)

# test output (labels) ko bhi tensor + column format me convert kiya
y_test = torch.tensor(y_test.values, dtype=torch.float32).view(-1,1)
```

```python
model = nn.Sequential(
    nn.Linear(2, 4),
    nn.ReLU(), #Hidden layers 

    nn.Linear(4, 1),
    nn.Sigmoid() # Output layer → Sigmoid (binary case)
)
print(model)
```

```python
import torch.optim as optim
import torch.nn as nn

# Binary classification ke liye loss
criterion = nn.BCELoss()

# Optimizer (weights update karega)
optimizer = optim.Adam(model.parameters(), lr=0.01)
```

```python
epochs = 500

for epoch in range(epochs):

    # 🔸 Forward pass (prediction nikalna)
    outputs = model(x_train)

    # 🔸 Loss calculate
    loss = criterion(outputs, y_train)

    # 🔸 Backpropagation
    optimizer.zero_grad()  # purane gradients hatao
    loss.backward()        # naye gradients calculate karo
    optimizer.step()       # weights update karo

    # 🔸 Progress print
    if (epoch+1) % 100 == 0:
        print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")
```
